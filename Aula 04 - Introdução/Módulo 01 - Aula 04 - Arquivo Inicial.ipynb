Referência do pyautogui: https://pyautogui.readthedocs.io/en/latest/quickstart.html

```python
# Passo a passo

# Passo 01 - Entrar no sistema da empresa (link no drive)
# Passo 02 - Navegar no sistema e encontrar a base de dados
# Passo 03 - Exportar a base de dados (baixar arquivo)
# Passo 04 - Calcular os indicadores (faturamento e quantidade de produtos vendidos)
# Passo 05 - Enviar informações por E-mail

# Instalação necessária de Bibliotecas (Após instalado, não precisa ser incluído novamente em outros projetos)

# pyautogui > Automações no PC
# pandas > Manipulação e Visualização de Dados
# openpyxl > Manipulação de bases de dados de Excel (.xlsx)

# FUNÇÕES pyautogui

# pyautogui.click - Clicar com mouse
# pyautogui.write - Escrever um texto
# pyautogui.press - Pressionar uma tecla específica (ENTER)
# pyautogui.hotkey - Combinação de teclas (CRTL+C)

# PRA A INSTALAÇÃO: Abrir terminal e executar: pip install pandas pyautogui openpyxl
```

```python
# Passo 01 - Entrar no sistema da empresa (link no drive)
    # Abrir Google Chrome

import pyautogui
import time

# Após o "." é chamado de Função em Python, executa uma ação na função chamada.

pyautogui.PAUSE = 1 # Função para guardar 1 segundo para cada ação da função pyautogui - Útil para evitar erros em processos muito rápidos

pyautogui.press('win')
pyautogui.write('chrome')
pyautogui.press('enter')

time.sleep(1) # Faz o código pausar NESTA ETAPA por um período de tempo

    # Clicar no usuário correto no Google Chrome
pyautogui.click(x=1229, y=484, clicks=2) # Inserir coordenadas do usuário e quantidade de cliques
time.sleep(1) # Faz o código pausar NESTA ETAPA por um período de tempo

    # Incluir o Link para acesso e aguardar
link = "https://drive.google.com/drive/folders/149xknr9JvrlEnhNWO49zPcw0PW5icxga" #Armazenar o link como uma variável - ELIMINA NECESSIDADE DE ESCREVER O LINK VÁRIAS VEZES
pyautogui.write(link)
time.sleep(1) # Faz o código pausar NESTA ETAPA por um período de tempo

    # ENTER e aguardar
pyautogui.press('enter')
time.sleep(2)

# Passo 02 - Navegar no sistema e encontrar a base de dados

    # Clicar no usuário correto no Google Chrome
pyautogui.click(x=-1558, y=440, clicks=1) # Inserir coordenadas do usuário e quantidade de cliques
time.sleep(10) # Faz o código pausar NESTA ETAPA por um período de tempo

# Passo 03 - Exportar a base de dados (baixar arquivo)
    # Clicar no arquivo
pyautogui.click(x=-1517, y=369, clicks=2) # Inserir coordenadas do usuário e quantidade de cliques
time.sleep(2) # Faz o código pausar NESTA ETAPA por um período de tempo
    # Clicar em 3 pontos
pyautogui.click(x=-1420, y=373, clicks=1) # Inserir coordenadas do usuário e quantidade de cliques
time.sleep(2) # Faz o código pausar NESTA ETAPA por um período de tempo
    # Clicar em Baixar
pyautogui.click(x=-1380, y=461, clicks=1) # Inserir coordenadas do usuário e quantidade de cliques
time.sleep(2) # Faz o código pausar NESTA ETAPA por um período de tempo
```

### Vamos agora ler o arquivo baixado para pegar os indicadores

- Faturamento
- Quantidade de Produtos

```python
# Passo 04 - Calcular os indicadores (faturamento e quantidade de produtos vendidos)
import pandas

    # Abrir a base de dados
caminho = r"C:\Users\Usuario\Downloads\Vendas - Dez.xlsx"    # O 'r' é uma função que define que o que vem escrito após ele deve ser lido literalmente. Em Python operadores como '\U' podem executar ações
tabela = pandas.read_excel(caminho)

    # Ver as informações da base de dados
display(tabela)     # A função 'display' é semelhante a função print, sendo mais indicadas para tabelas por trazer formatado visualmente

    # Somar o faturamento de todos os produtos - Somar a coluna 'Valor Final'
faturamento = tabela["Valor Final"].sum()
    # Somar a quantidade de produtos - Somar a coluna 'Quantidade'
qtde_produtos = tabela["Quantidade"].sum()

print(faturamento)
print(qtde_produtos)
```
