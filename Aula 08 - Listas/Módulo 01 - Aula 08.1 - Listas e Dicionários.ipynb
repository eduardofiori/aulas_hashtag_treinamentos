#### Em Python, listas e dicionários são estruturas fundamentais para armazenar e organizar dados. 

Resumidamente: Listas são mutáveis (podem ser alteradas);Dicionários armazenam pares de chave-valor.

### Listas

```python
frutas = ["maçã", "banana", "laranja"]
frutas.append("uva")  # Adiciona um item
frutas[0] = "abacaxi" # Altera o primeiro item
print(frutas)
```

### Dicionários

```python
usuario = {
    "nome": "Ana",
    "idade": 28,
    "cidade": "Ribeirão Preto"
}
print(usuario["nome"]) # Imprime "Ana"
```
