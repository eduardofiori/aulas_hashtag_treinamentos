## 1. Faturamento do Melhor e do Pior Mês do Ano

Qual foi o valor de vendas do melhor mês do Ano?
E valor do pior mês do ano?

```python
meses = ['jan', 'fev', 'mar', 'abr', 'mai', 'jun', 'jul', 'ago', 'set', 'out', 'nov', 'dez']
vendas_1sem = [25000, 29000, 22200, 17750, 15870, 19900]
vendas_2sem = [19850, 20120, 17540, 15555, 49051, 9650]

#Unir os semestres em uma única lista
vendas_ano = vendas_1sem + vendas_2sem

```

## 2. Continuação

Agora relacione as duas listas para printar 'O melhor mês do ano foi {} com {} vendas' e o mesmo para o pior mês do ano.

Calcule também o faturamento total do Ano e quanto que o melhor mês representou do faturamento total.

Obs: Para o faturamento total, pode usar a função sum(lista) que soma todos os itens de uma lista

```python
#Calcular melhor mês
vendas_max = max(vendas_ano)
mes_max = vendas_ano.index(vendas_max)

#Calcular pior mês
vendas_min = min(vendas_ano)
mes_min = vendas_ano.index(vendas_min)

#% de representação do melhor mês
vendas_max_percent = vendas_max / sum(vendas_ano)

#Resultados
print('O melhor mês de vendas foi: {}, com R$ {:,.2f} em total de vendas, representando {:,.1%}. O pior mês de vendas foi: {}, com R$ {:,.2f} em total de vendas.'.format(meses[mes_max],vendas_ano[mes_max],vendas_max_percent,meses[mes_min],vendas_ano[mes_min]))
```

## 3. Crie uma lista com o top 3 valores de vendas do ano (sem fazer "no olho")

```python
#Top1
vendas_max = max(vendas_ano)
vendas_top3 = []
vendas_top3.append(vendas_max)
vendas_ano.remove(vendas_max)

#Top2
vendas_max = max(vendas_ano)
vendas_top3.append(vendas_max)
vendas_ano.remove(vendas_max)

#Top3
vendas_max = max(vendas_ano)
vendas_top3.append(vendas_max)
vendas_ano.remove(vendas_max)

#Lista Top 3
print(vendas_top3)
```
