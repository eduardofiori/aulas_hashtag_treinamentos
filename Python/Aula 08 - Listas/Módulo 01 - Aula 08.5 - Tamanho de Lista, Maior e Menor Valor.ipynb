## Maior e Menor Valor

maior = max(lista)

menor = min(lista)

```python
vendas = [1000, 1500, 15000, 270, 900, 100, 1200]
mais_vendas = max(vendas)
menos_vendas = min(vendas)
print('O produto mais vendido teve {} unidades vendidas. O Produto menos vendidos teve {} unidades vendidas'.format(mais_vendas,menos_vendas))
```

- Qual o item mais vendido?
- Qual o item menos vendido?

```python
vendas = [1000, 1500, 15000, 270, 900, 100, 1200]
mais_vendas = max(vendas)
i_mais_vendas = vendas.index(mais_vendas)


menos_vendas = min(vendas)
i_menos_vendas = vendas.index(menos_vendas)

print('O produto mais vendido teve foi o {} e teve {} unidades vendidas. O Produto menos vendidos foi o {} e teve {} unidades vendidas'.format(i_mais_vendas,mais_vendas,i_menos_vendas,menos_vendas))
```
