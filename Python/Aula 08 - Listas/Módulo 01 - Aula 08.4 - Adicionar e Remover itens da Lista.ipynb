# Adicionar e Remover itens de uma lista

Adicionar:<br>
lista.append(item)

Remover:<br>
item_removido = lista.pop(indice)<br>
lista.remove(item)

Digamos que você está construindo o controle de produtos da Apple.<br>
E a Apple lançou o IPhone 11 e irá tirar dos seus estoques o IPhone X

```python
produtos = ['apple tv', 'mac', 'iphone x', 'IPad', 'apple watch', 'mac book', 'airpods']

#Adicionar 

produtos.append('iphone 11')
print(produtos)

#Remover - Podemos usar '.remove' ou '.pop'. Tem a mesma função, porém o segundo é usado consultando o índice da lista, além de poder ser transformado em variável
produtos.remove('iphone x')
print(produtos)

item_removido = produtos.pop(2)
print('Removemos o item {} da lista'.format(item_removido))
print(produtos)

```

## Existem 2 formas de tratar o erro:

1. Criar um if para evitar que ele aconteça

2. Esperar que ele possa acontecer e tratar caso o erro aconteça com:

```python
consulta_produto = input('Consulta de produto: ')
try:

    produtos.remove(consulta_produto)
    print(produtos)
except:
    print('O produto {} não foi encontrado na lista'.format(consulta_produto))
```
