```python
faturamento = 1000
custo = 500
lucro = faturamento - custo

#Não podemos misturar tipos de variáveis em concatenação e cálculos, assim precisamos transformar o 'faturamento' em string - str()
print('O Faturamento da Loja foi ' + str(faturamento))
print('O Custo da loja foi ' + str(custo))
print('Assim, o Lucro da Loja foi de ' + str(custo) +' reais') 

#Podemos colocar em apenas uma linha
#O operador "/n/n" insere um parágrafo entre textos ao concatenar
print("\n\n" + 'O Faturamento da Loja foi ' + str(faturamento) + 'O Custo da Loja foi ' + str(custo) + 'Assim, o Lucro da Loja foi de ' + str(custo) +' reais')
```

```python
faturamento = 1000
custo = 500
lucro = faturamento - custo

# '{}.format(Nome da Variavel)' utilizado para não precisar formatar o tipo de variável, além de facilitar a chamada de múltiplas variáveis
print('O faturamento da loja foi de {}'.format(faturamento))

#Ou podemos utilizar para o texto completo
print("\n\n" + 'O faturamento da loja foi de {}. O Custo da Loja foi de {}. Assim, o Lucro foi de {}'.format(faturamento,custo,lucro))
```
