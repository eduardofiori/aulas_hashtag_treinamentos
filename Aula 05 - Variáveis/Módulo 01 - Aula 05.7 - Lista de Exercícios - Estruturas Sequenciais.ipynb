#### 12. Faça um Programa que pergunte quanto você ganha por hora e o número de horas trabalhadas no mês.

```python
#Solicitar salário por hora e horas trabalhas
salario_hora = float(input("Qual seu salário médio por hora? "))
horas_trabalhadas = int(input("Qual o total de horas trabalhadas no mês? "))

```

#####  Calcule o salário bruto (horas * salario por hora)

```python
#Calcular salário total
salario = horas_trabalhadas * salario_hora
#Mostrar resultado
print(f"Seu salário total no mês é de R$ {salario:,.2f}")
```

##### Calcule o desconto do IR (11% do salário bruto)

```python
#Desconto do IR
desconto_ir = salario * 0.11
#Mostrar resultado
print(f"Seu desconto de IR é de R$ {desconto_ir:,.2f}")
```

##### Calcule o desconto do INSS (8% do salário bruto)

```python
#Desconto do IR
desconto_inss = salario * 0.08
#Mostrar resultado
print(f"Seu desconto de INSS é de R$ {desconto_inss:,.2f}")
```

##### Calcule o desconto do sindicato (5% do salário bruto)

```python
#Desconto do Sindicato
desconto_sindicato = salario * 0.05
#Mostrar resultado
print(f"Seu desconto do Sindicato é de R$ {desconto_sindicato:,.2f}")
```

##### Calcule o salário líquido (salário bruto - descontos)

```python
#Calculo de Salario Líquido
salario_liquido = salario - desconto_inss - desconto_ir - desconto_sindicato
#Mostrar resultado
print(f"Seu Salário Líquido é de R$ {salario_liquido:,.2f}")
```

#### 14. Faça um programa que peça o tamanho de um arquivo para download (em MB) e a velocidade de um link de Internet (em Mbps), calcule e informe o tempo aproximado de download do arquivo usando este link (em minutos).

Detalhe: MB significa megabyte, Mb (com b minúsculo) significa megabit. Um megabit é 1/8 de um megabyte. 

```python
#Solicitar Tamanho do Arquivo (MB) e converter (Mb)
tamanho_arquivo = float(input("Qual o tamanho do arquivo em MB? "))
tamanho_arquivo_convertido = tamanho_arquivo * 8

#Solciitar Velocidade de Upload (Mbps)
velocidade_upload = float(input("Qual a Velocidade de Upload em Mbps? "))

#Calculo de Tempo de Download em minutos (m)
tempo_download =  tamanho_arquivo_convertido / velocidade_upload
tempo_download_min = tempo_download / 60

#Resultado
print(f"A Velocidade de Download do Arquivod de {tamanho_arquivo:,.2f}MB será de {tempo_download_min:,.1f} minutos")
```
