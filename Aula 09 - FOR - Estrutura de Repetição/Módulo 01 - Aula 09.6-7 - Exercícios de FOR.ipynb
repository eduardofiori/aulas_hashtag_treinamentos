## 2. Análise de Vendas

Nesse exercício vamos fazer uma "análise simples" de atingimento de Meta.

Temos uma lista com os vendedores e os valores de vendas e queremos identificar (printar) quais os vendedores que bateram a meta e qual foi o valor que eles venderam.

```python
meta = 10000
vendas = [
    ['João', 15000],
    ['Julia', 27000],
    ['Marcus', 9900],
    ['Maria', 3750],
    ['Ana', 10300],
    ['Alon', 7870],
]
#Seu código aqui

for venda in vendas:
    if int(venda[1]) >= meta:
        print("O vendedor(a) {} 🟢 ATINGIU A META e vendeu R$ {:,.0f}".format(venda[0],venda[1]))
    else:
        print("O vendedor(a) {} 🔴 NÃO ATINGIU A META e vendeu R$ {:,.0f}".format(venda[0],venda[1]))
```

```python
meta = 10000
vendas = [
    ['João', 15000],
    ['Julia', 27000],
    ['Marcus', 9900],
    ['Maria', 3750],
    ['Ana', 10300],
    ['Alon', 7870],
]

bateram_meta = []
nao_bateram_meta = []

for venda in vendas:
    if int(venda[1]) >= meta:
        bateram_meta.append(venda) #Adiciona a venda na lista 'bateram_meta'
    else:
        nao_bateram_meta.append(venda) #Adiciona a venda na lista 'nao_bateram_meta'

print("=== Vendedores que 🟢 ATINGIRAM a meta ===")
for venda in bateram_meta:
    print("O vendedor(a) {} vendeu R$ {:,.0f}".format(venda[0], venda[1]))

print("\n=== Vendedores que 🔴 NÃO atingiram a meta ===") #'\n' inserido para quebrar a linha de texto (parágrafo)
for venda in nao_bateram_meta:
    print("O vendedor(a) {} vendeu apenas R$ {:,.0f}".format(venda[0], venda[1]))

total = len(vendas)
qtde_bateu = len(bateram_meta)
qtde_nao_bateu = len(nao_bateram_meta)

print(f"\nResumo: {qtde_bateu}/{total} bateram a meta ({qtde_bateu/total:.1%})")
```
